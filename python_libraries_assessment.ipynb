```python
import pandas as pd

df = pd.DataFrame({
    'order_id': [1, 2, 3, 4, 5],
    'restaurant_name': ['A', 'A', 'B', 'B', 'C'],
    'city': ['Chennai', 'Chennai', 'Mumbai', 'Mumbai', 'Delhi'],
    'order_value': [200, 300, 150, 250, 400],
    'delivery_time_mins': [30, 40, 25, 35, 45],
    'rating': [4, 5, 3, 4, 5]
})

result = df.groupby('restaurant_name').agg({
    'order_value': 'mean',
    'delivery_time_mins': 'mean',
    'rating': 'mean'
})

result
```
